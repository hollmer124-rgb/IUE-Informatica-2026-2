Crear y cargar una lista con 5 enteros. Implementar un algoritmo que identifique el mayor valor de la lista.

```python
lista=[]
for x in range(5):
    valor=int(input("Ingrese valor:"))
    lista.append(valor)

mayor=lista[0]
for x in range(1,5):
    if lista[x]>mayor:
        mayor=lista[x]

print("Lista completa")
print(lista)
print("Mayor de la lista")
print(mayor)
```

Crear y cargar una lista con 5 enteros por teclado. Implementar un algoritmo que identifique el menor valor de la lista y la posición donde se encuentra.

Cargar una lista con 5 elementos enteros. Imprimir el mayor y un mensaje si se repite dentro de la lista (es decir si dicho valor se encuentra en 2 o más posiciones en la lista)

```python
lista=[]
for x in range(5):
    valor=int(input("Ingrese valor:"))
    lista.append(valor)

mayor=lista[0]
for x in range(1,5):
    if lista[x]>mayor:
        mayor=lista[x]

print("Lista completa")
print(lista)
print("Mayor de la lista")
print(mayor)

# contamos cuantas veces se encuentra el mayor en la lista
cantidad=0
for x in range(5):
    if lista[x]==mayor:
        cantidad=cantidad+1
if cantidad>1:
    print("El mayor se repite en la lista")
```

Desarrollar un programa que permita cargar 5 nombres de personas y sus edades respectivas. Luego de realizar la carga por  teclado de todos los datos imprimir los nombres de las personas mayores de edad (mayores o iguales a 18 años)

```python
nombres=[]
edades=[]
for x in range(5):
    nom=input("Ingrese el nombre de la persona:")
    nombres.append(nom)
    ed=int(input("Ingrese la edad de dicha persona:"))
    edades.append(ed)

print("Nombre de las personas mayores de edad:")
for x in range(5):
    if edades[x]>=18:
        print(nombres[x])
```

Crear y cargar dos listas con los nombres de 5 productos en una y sus respectivos  precios en otra. Definir dos listas paralelas. Mostrar cuantos productos tienen un precio mayor al primer producto ingresado.

```python
productos=[]
precios=[]
for x in range(5):
    nom=input("Ingrese el nombre del producto:")
    productos.append(nom)
    pre=int(input("Ingrese el precio de dicho producto:"))
    precios.append(pre)

cantidad=0
for x in range(1,5):
    if precios[x]>precios[0]:
        cantidad=cantidad+1

print("Cantidad de productos con un precio mayor al primer producto ingresado")
print(cantidad)
```

En un curso de 4 alumnos registraron las notas de sus examenes y se deben procesar de acuerdo a lo sieguiente: a) ingresar nombre y nota de cada alumno(almacenar los datos en dos listas paralelas b) realizar un listado que muestre los nombres, notas y condicion del alumno. En la condicion, colocar "Muy bueno" si la nota es mayor o igual a 8,"Bueno" si la nota esta entre 4 y 7, y colocar "Insuficiente" si la nota es inferior a 4. c)j impriir cuantos alumnos tienen la leyenda "muy bueno
"

```python
alumnos = []
notas = []
for x in range(4):
  nom = input("Ingrese el nombre del alumno: ")
  alumnos.append(nom)
  nota = float(input("Ingrese la nota del alumno: "))
  notas.append(nota)
muybueno = 0
for x in range(4):
  print(f"Alumno: {alumnos[x]}")
  print(f"Nota: {notas[x]}")
  if notas[x] >= 8:
    print("Condicion: Muy Bueno")
    muybueno += 1
  elif notas[x] >= 4 and notas[x] <= 7:
    print("Condicion: Bueno")
  elif notas[x] < 4:
    print("Condicion: Insuficiente")

print(f"La cantidad de alumnos con Muy bueno es: {muybueno}")
```
