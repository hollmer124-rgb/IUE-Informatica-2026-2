Ejercicio Libre
Software de cajero: Programa con lista en el que se asigna un nombre de usuario, contraseña, saldo inicial de  100.000 $, con login, y la opcion de retirar y consignar al saldo


```python
from decimal import Decimal
#Lista con datos del usuario
usuario = ["Mike", "Messi123", Decimal("100000.0")]
#Login
while True:
  nombre = input("Ingrese el nombre de usuario: ")
  contraseña = str(input("Ingrese su contraseña: "))
  #Validacion de datos
  if nombre == usuario[0] and contraseña == usuario[1]:
    print("Inicio de sesion exitoso... \nBienvenido al Sistema")
    break
  else:
    print("Nombre o contraseña invalidos, vuelva a intentarlo")

#Bucle principal del sistema
while True:
  print("1. Consignar")
  print("2. Retirar")
  print("3. Salir")
  opcion = input("Ingrese el numero de la opcion seleccionada:  ")

  #Consignar
  if opcion == "1":
    while True:
      try:
        consignar = Decimal(input("Ingrese la cantidad que quiere consignar: "))
        if consignar > 0:
          usuario[2] += consignar
          print(f"Su saldo actual es: {usuario[2]} ")
          break
        else:
          print("Error, Ingrese una cantidad valida...")
      except ValueError:
        print("Error, ingrese una cantidad valida...")

  #Retirar
  elif opcion == "2":
    while True:
      try:
        retirar = Decimal(input("Ingrese la cantidad que desea retirar: "))
        if retirar <= usuario[2] and retirar > 0:
          usuario[2] -= retirar
          print(f"Su saldo actual es: {usuario[2]} ")
          break

        else:
          print("Error, ingrese una cantidad valida...")

      except ValueError:
        print("Error, Ingrese una cantidad valida")

  #Salir

  elif opcion == "3":
    print("Gracias por usar el Sistema...")
    break
  else:
    print("Opcion invalida, vuelva intentarlo...")

```
