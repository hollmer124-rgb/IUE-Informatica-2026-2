Sistema de Gestión de Inventario para una Tienda de Perfumes

```python
# Listas para almacenar la información de los perfumes
nombres = []
precios = []
cantidades = []

# Variable para controlar el menú
opcion = 0


while True:

    print("\n--- TIENDA DE PERFUMES ---")
    print("1. Cargar perfumes")
    print("2. Ver inventario completo")
    print("3. Ver reporte de stock crítico")
    print("4. Ver valor total del inventario")
    print("5. Salir")

    opcion = input("Seleccione una opción: ")


    # OPCIÓN 1: Cargar perfumes
    if opcion == "1":

        # Se instanciaron las listas vacias para cargar un nuevo inventario
        nombres = []
        precios = []
        cantidades = []

        cantidad_productos = int(input("¿Cuántos perfumes desea registrar?: "))

        for i in range(cantidad_productos):

            print("\nPerfume", i + 1)

            nombre = input("Nombre del perfume: ")
            precio = float(input("Precio del perfume: "))
            cantidad = int(input("Cantidad en stock: "))

            # Guardamos cada dato en su respectiva lista
            nombres.append(nombre)
            precios.append(precio)
            cantidades.append(cantidad)

        print("\nPerfumes registrados correctamente.")


    # OPCIÓN 2: Ver inventario completo
    elif opcion == "2":

        if len(nombres) == 0:
            print("\nNo hay perfumes registrados.")

        else:
            print("\n--- INVENTARIO COMPLETO ---")

            for i in range(len(nombres)):

                print("\nPerfume:", nombres[i])
                print("Precio: $", precios[i])
                print("Cantidad en stock:", cantidades[i])

            # Buscar el perfume con el precio más alto
            posicion_mayor = 0

            for i in range(1, len(precios)):

                if precios[i] > precios[posicion_mayor]:
                    posicion_mayor = i

            print("\nPerfume con el precio más alto:")
            print(nombres[posicion_mayor])
            print("Precio: $", precios[posicion_mayor])


    # OPCIÓN 3: Ver reporte de stock crítico
    elif opcion == "3":

        if len(nombres) == 0:
            print("\nNo hay perfumes registrados.")

        else:
            print("\n--- PRODUCTOS CON STOCK CRÍTICO ---")

            hay_stock_critico = False

            for i in range(len(nombres)):

                # Un producto tiene stock crítico si es menor a 5
                if cantidades[i] < 5:

                    print(nombres[i], "- Cantidad:", cantidades[i])

                    hay_stock_critico = True

            if hay_stock_critico == False:
                print("No hay perfumes con stock crítico.")


    # OPCIÓN 4: Ver valor total del inventario
    elif opcion == "4":

        if len(nombres) == 0:
            print("\nNo hay perfumes registrados.")

        else:
            valor_total = 0

            for i in range(len(nombres)):

                # Precio multiplicado por la cantidad en stock
                valor_total = valor_total + (precios[i] * cantidades[i])

            print("\n--- VALOR TOTAL DEL INVENTARIO ---")
            print("El valor total es: $", valor_total)


    # OPCIÓN 5: Salir
    elif opcion == "5":

        print("\nGracias por usar el sistema de la tienda de perfumes.")
        break


    # Opción inválida
    else:

        print("\nOpción no válida. Intente nuevamente.")
```
