Método de bisección

```python
import sympy as sp

def biseccion(funcion_str, a, b, tol=1e-6, max_iter=100):
    # Definir variable y función
    x = sp.Symbol('x')
    f = sp.sympify(funcion_str)

    # Verificar que hay un cambio de signo
    fa = f.subs(x, a)
    fb = f.subs(x, b)
    if fa * fb > 0:
        raise ValueError("La función no cambia de signo en el intervalo dado.")

    print(f"{'Iter':<5} {'a':<12} {'b':<12} {'m':<12} {'f(m)':<12}")
    for i in range(1, max_iter + 1):
        m = (a + b) / 2
        fm = f.subs(x, m)

        print(f"{i:<5} {a:<12.6f} {b:<12.6f} {m:<12.6f} {fm:<12.6f}")

        if abs(fm) < tol or (b - a) / 2 < tol:
            return float(m)

        if fa * fm < 0:
            b = m
            fb = fm
        else:
            a = m
            fa = fm

    raise ValueError("No se alcanzó la tolerancia en el número máximo de iteraciones.")
```

```python
  raiz = biseccion("x**3 - x - 2", a=1, b=2, tol=1e-6, max_iter=50)
  print(f"\nLa raíz aproximada es: {raiz}")
```

Método de la falsa posición

```python
import sympy as sp

def falsa_posicion(funcion_str, a, b, tol=1e-6, max_iter=100):
    """
    Método de Falsa Posición (Regula Falsi) para encontrar una raíz de f(x)=0.
    - funcion_str: cadena con la función, p.ej. "x**3 - x - 2"
    - a, b: extremos iniciales del intervalo [a,b] con f(a)*f(b) < 0
    - tol: tolerancia para |f(c)| o longitud de intervalo
    - max_iter: máximo número de iteraciones
    """
    # Definir variable y función
    x = sp.Symbol('x')
    f_expr = sp.sympify(funcion_str)
    f = sp.lambdify(x, f_expr, 'math')

    fa = f(a)
    fb = f(b)

    # Verificar cambio de signo
    if fa * fb > 0:
        raise ValueError("La función no cambia de signo en el intervalo dado.")

    print(f"{'Iter':<5} {'a':<12} {'b':<12} {'c':<12} {'f(c)':<12} {'|b-a|':<12}")
    for i in range(1, max_iter + 1):
        # Fórmula de Falsa Posición
        c = b - fb * (b - a) / (fb - fa)
        fc = f(c)

        print(f"{i:<5} {a:<12.6f} {b:<12.6f} {c:<12.6f} {fc:<12.6f} {abs(b-a):<12.6f}")

        # Criterios de parada
        if abs(fc) < tol or abs(b - a) < tol:
            return float(c)

        # Actualización del intervalo
        if fa * fc < 0:
            b = c
            fb = fc
        else:
            a = c
            fa = fc

    raise ValueError("No se alcanzó la tolerancia en el número máximo de iteraciones.")
```

```python
raiz = falsa_posicion("x**3 - x - 2", a=1.0, b=2.0, tol=1e-6, max_iter=50)
print(f"\nLa raíz aproximada es: {raiz}")
```

Método de Newton

```python
import sympy as sp

def newton(funcion_str, x0, tol=1e-6, max_iter=100):
    """
    Método de Newton-Raphson para resolver f(x)=0.
    - funcion_str: cadena con la función, p.ej. "x**3 - x - 2"
    - x0: aproximación inicial
    - tol: tolerancia para |f(x)| o |Δx|
    - max_iter: máximo de iteraciones
    """
    # Símbolo y expresiones
    x = sp.Symbol('x')
    f_expr = sp.sympify(funcion_str)
    df_expr = sp.diff(f_expr, x)

    # Funciones numéricas (rápidas)
    f = sp.lambdify(x, f_expr, 'math')
    df = sp.lambdify(x, df_expr, 'math')

    print(f"{'Iter':<5} {'xk':<12} {'f(xk)':<12} {'df(xk)':<12} {'xk+1':<12} {'|Δx|':<12}")

    for i in range(1, max_iter + 1):
        fx = f(x0)
        dfx = df(x0)

        if dfx == 0:
            raise ValueError("Derivada nula en el punto actual; no se puede continuar con Newton.")

        x1 = x0 - fx / dfx
        f1 = f(x1)
        dx = abs(x1 - x0)

        print(f"{i:<5} {x0:<12.6f} {fx:<12.6f} {dfx:<12.6f} {x1:<12.6f} {f1:<12.6f} {dx:<12.6f}")

        # Criterios de parada
        if abs(f1) < tol or dx < tol:
            return float(x1)

        x0 = x1

    raise ValueError("No se alcanzó la tolerancia en el número máximo de iteraciones.")
```

```python
raiz = newton("x**3 - x - 2", x0=1.5, tol=1e-6, max_iter=50)
print(f"\nLa raíz aproximada es: {raiz}")
```

Metodo de la secante

```python
import sympy as sp

def secante(funcion_str, x0, x1, tol=1e-6, max_iter=100):
    """
    Método de la secante para encontrar una raíz de f(x)=0.
    - funcion_str: cadena con la función, p.ej. "x**3 - x - 2"
    - x0, x1: aproximaciones iniciales
    - tol: tolerancia para |f(x)| o |x_{k+1} - x_k|
    - max_iter: máximo número de iteraciones
    """
    # Definir variable y función
    x = sp.Symbol('x')
    f_expr = sp.sympify(funcion_str)
    # Convertir a función numérica para acelerar evaluaciones
    f = sp.lambdify(x, f_expr, 'math')

    f0 = f(x0)
    f1 = f(x1)

    print(f"{'Iter':<5} {'x0':<12} {'x1':<12} {'x2':<12} {'f(x2)':<12} {'|Δx|':<12}")
    for i in range(1, max_iter + 1):
        denom = (f1 - f0)
        if denom == 0:
            raise ValueError("División por cero en la fórmula de la secante (f(x1) == f(x0)).")

        x2 = x1 - f1 * (x1 - x0) / denom
        f2 = f(x2)
        dx = abs(x2 - x1)

        print(f"{i:<5} {x0:<12.6f} {x1:<12.6f} {x2:<12.6f} {f2:<12.6f} {dx:<12.6f}")

        # Criterios de parada: valor cercano a cero o cambio en x pequeño
        if abs(f2) < tol or dx < tol:
            return float(x2)

        # Actualizar para la siguiente iteración
        x0, x1 = x1, x2
        f0, f1 = f1, f2

    raise ValueError("No se alcanzó la tolerancia en el número máximo de iteraciones.")
```

```python
raiz = secante("x**3 - x - 2", x0=1.0, x1=2.0, tol=1e-6, max_iter=50)
print(f"\nLa raíz aproximada es: {raiz}")
```
